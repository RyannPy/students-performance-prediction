# Students Peformance Analytics and Predict

Studi Kasus: Analisis Prediktif Faktor Sosio-Akademik Terhadap Capaian Pembelajaran Siswa

Dataset ini merekam berbagai variabel yang mencakup performa akademik histori, indikator demografis, latar belakang sosio-ekonomi, serta pola perilaku harian siswa. Data ini mencerminkan kompleksitas faktor-faktor sistemik dan individual yang saling bertautan dalam memengaruhi keberhasilan studi seorang peserta didik.

Tujuan Utama (Goal):

Membangun model analisis prediktif yang kuat untuk mengestimasi nilai akhir siswa secara akurat. Melalui pendekatan ini, institusi dapat melakukan intervensi dini secara preventif terhadap siswa yang teridentifikasi berisiko mengalami penurunan performa, sekaligus mengevaluasi efektivitas kebijakan kurikulum yang sedang berjalan.

Objektif (Objectives):

- Menemukan dan mengukur korelasi kausalitas antara faktor non-akademik (seperti absensi, dukungan keluarga, dan tingkat stres) dengan hasil belajar kognitif siswa.

- Mengembangkan algoritma machine learning yang mampu menggeneralisasi tren performa tanpa terjebak pada bias demografis, sehingga prediksi tetap objektif dan adil (fairness in AI).

- Menguraikan pola tersembunyi (hidden patterns) dalam dataset untuk memahami mengapa kelompok siswa tertentu menunjukkan tren performa yang stagnan atau menurun.
